```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
steam = pd.read_csv('C:\\Users\\janek\\Documents\\Steamify\\table_csvs\\steam\\game.csv')
steam.head()
```

```python
scaler = MinMaxScaler(feature_range=(0,100))
numeric_columns = ['positive_negative_ratings_ratio','median_playtime','total_ratings']
scaled_steam = scaler.fit_transform(steam[numeric_columns])
scaled_steam = pd.DataFrame(scaled_steam, columns=numeric_columns, index=steam.index)
scaled_steam.describe()
```

```python
new_steam=steam.drop(columns=numeric_columns)
new_steam=new_steam.merge(scaled_steam, left_index=True, right_index=True)
new_steam.head()
```

```python
new_steam[['game_id','game_name','required_age','positive_negative_ratings_ratio','median_playtime','total_ratings','price']].to_csv('C:\\Users\\janek\\Documents\\Steamify\\table_csvs\\steam\\game.csv',index=False)
```

```python
spotify = pd.read_csv('C:\\Users\\janek\\Documents\\Steamify\\table_csvs\\spotify\\song.csv')
spotify.head()
```

```python
scaler = MinMaxScaler(feature_range=(0,100))
numeric_columns = ['duration','popularity','danceability','energy','acousticness','instrumentalness','valence']
scaled_spotify = scaler.fit_transform(spotify[numeric_columns])
scaled_spotify = pd.DataFrame(scaled_spotify, columns=numeric_columns, index=spotify.index)
scaled_spotify.head()
```

```python
new_spotify=spotify.drop(columns=numeric_columns)
new_spotify=new_spotify.merge(scaled_spotify, left_index=True, right_index=True)
new_spotify.head()
```

```python
new_spotify[['song_id','song_name','album','duration','popularity','explicit','danceability','energy','acousticness','instrumentalness','valence']].to_csv('C:\\Users\\janek\\Documents\\Steamify\\table_csvs\\spotify\\song.csv',index=False)
```
